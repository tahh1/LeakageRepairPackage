```python
# In[1]:


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# In[2]:


iris_df = pd.read_csv('../input/iris/Iris.csv')
iris_df.head()
```

```python
# In[3]:


iris_df.shape
```

```python
# In[4]:


iris_df = pd.get_dummies(iris_df, columns=['Species'])
iris_df.head()


# One-hot labels by get_dummies() method
```

```python
# In[5]:


X = iris_df.values[:, 1:5]
y = iris_df.values[:, 5:8]
```

```python
# In[6]:


def fit_normalizer(array):
    # min/max statistics are computed on the training data only
    return array.min(axis=(0, 1)), array.max(axis=(0, 1))

def normalize(array, arr_min, arr_max):
    return (array - arr_min) / (arr_max - arr_min)
```

```python
# In[7]:


from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
```

```python
# In[8]:


# normalize dataset between (0, 1) using statistics from the training split only
arr_min, arr_max = fit_normalizer(X_train)
X_train = normalize(X_train, arr_min, arr_max)
X_test = normalize(X_test, arr_min, arr_max)


# # Build NN with Keras
```

```python
# In[9]:


from keras.models import Sequential 
from keras.layers import Dense 
from keras.optimizers import Adam 
```

```python
# In[10]:


def create_network():
    model = Sequential()
    model.add(Dense(7, input_shape=(4,), activation='relu')) 
    model.add(Dense(8, activation='relu')) 
    model.add(Dense(5, activation='relu'))
    model.add(Dense(3, activation='softmax')) #3 classes need 3 nodes at output layer
        
    model.compile(loss='categorical_crossentropy', optimizer=Adam(), metrics=['accuracy']) 
    return model
```

```python
# In[11]:


model = create_network()
```

```python
# In[12]:


# hold out part of the training data for validation; the test set is kept unseen during training
X_tr, X_val, y_tr, y_val = train_test_split(X_train, y_train, test_size=0.1, random_state=42)
results = model.fit(X_tr, y_tr, epochs=10, batch_size=8, validation_data=(X_val, y_val))
```

```python
# In[13]:


plt.figure(figsize=(5, 5))
plt.title("Learning curve")
plt.plot(results.history["loss"], label="loss")
plt.plot(results.history["val_loss"], label="val_loss")
plt.plot( np.argmin(results.history["val_loss"]), np.min(results.history["val_loss"]), marker="x", color="r", label="best model")
plt.xlabel("Epochs")
plt.ylabel("log_loss")
plt.legend();
```
