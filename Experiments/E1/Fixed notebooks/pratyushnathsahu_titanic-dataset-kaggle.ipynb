```python
# In[2]:


train_df=pd.read_csv("../input/train.csv")
test_df=pd.read_csv("../input/test.csv")
```

```python
# In[3]:


print((train_df.info()))
print(("--"*20))
print((test_df.info()))
```

```python
# In[4]:


X=train_df.drop("Survived",axis=1)
Y=train_df["Survived"]
```

```python
# In[5]:


combined_df=pd.concat([X,test_df])
```

```python
# In[6]:


print((combined_df.info()))


# # Data Cleaning
# * Correct
# * Complete
# * Create
# * Convert
```

```python
# In[7]:


from sklearn.impute import SimpleImputer

Age_imputer=SimpleImputer(strategy="median")
Fare_imputer=SimpleImputer(strategy="median")
Embarked_imputer=SimpleImputer(strategy="most_frequent")

# Fit imputers on the training rows only (first 891 rows), then apply to all rows,
# so that statistics from the test set do not leak into preprocessing.
n_train=len(train_df)
combined_df=combined_df.reset_index(drop=True)
train_part=combined_df.iloc[:n_train]

Age_imputer.fit(train_part[["Age"]])
Fare_imputer.fit(train_part[["Fare"]])
Embarked_imputer.fit(train_part[["Embarked"]])

combined_df["Age"]=Age_imputer.transform(combined_df[["Age"]]).ravel()
combined_df["Fare"]=Fare_imputer.transform(combined_df[["Fare"]]).ravel()
combined_df["Embarked"]=Embarked_imputer.transform(combined_df[["Embarked"]]).ravel()
```

```python
# In[8]:


combined_df.drop(["Cabin"],axis=1,inplace=True)
```

```python
# In[9]:


combined_df["Family_size"]=combined_df["SibSp"]+combined_df["Parch"]+1
combined_df["Is_alone"]=int(0)
combined_df.loc[combined_df["Family_size"]==1,"Is_alone"]=int(1)
```

```python
# In[10]:


combined_df['Title']=combined_df['Name'].str.split(",", expand=True)[1].str.split(".", expand=True)[0]
```

```python
# In[11]:


# Derive bin edges from the training rows only, then apply them to all rows
_, fare_edges=pd.qcut(combined_df["Fare"].iloc[:n_train],4,retbins=True)
fare_edges[0], fare_edges[-1]=-np.inf, np.inf
combined_df["Fare_bin"]=pd.cut(combined_df["Fare"],fare_edges)

_, age_edges=pd.cut(combined_df["Age"].iloc[:n_train].astype(int),4,retbins=True)
age_edges[0], age_edges[-1]=-np.inf, np.inf
combined_df["Age_bin"]=pd.cut(combined_df["Age"].astype(int),age_edges)
```

```python
# In[12]:


from sklearn.preprocessing import LabelEncoder

# Fit each encoder on the training rows only; map unseen test categories to -1
def encode_train_only(col):
    enc=LabelEncoder()
    enc.fit(col.iloc[:n_train].astype(str))
    mapping={c:i for i,c in enumerate(enc.classes_)}
    return col.astype(str).map(mapping).fillna(-1).astype(int)

combined_df["Sex_encode"]=encode_train_only(combined_df["Sex"])
combined_df["Age_encode"]=encode_train_only(combined_df["Age_bin"])
combined_df["Fare_encode"]=encode_train_only(combined_df["Fare_bin"])
combined_df["Title_encode"]=encode_train_only(combined_df["Title"])
combined_df["Embarked_encode"]=encode_train_only(combined_df["Embarked"])
```

```python
# In[13]:


combined_df.head()
```

```python
# In[14]:


# Features to be used
df_final=combined_df[["Pclass","Family_size","Is_alone","Sex_encode","Age_encode","Fare_encode","Embarked_encode"]]
X_train=df_final.iloc[:n_train]
X_test=df_final.iloc[n_train:]
```

```python
# In[15]:


X_train.head()


# # Model Creation
```

```python
# In[16]:


from sklearn import ensemble, neural_network , neighbors , svm , linear_model ,naive_bayes ,tree , gaussian_process,discriminant_analysis
from xgboost import XGBClassifier, XGBRFClassifier
```

```python
# In[17]:


#Machine Learning Algorithms
MLA=[
    #Ensemble Methods
    ensemble.AdaBoostClassifier(),
    ensemble.BaggingClassifier(),
    ensemble.ExtraTreesClassifier(),
    ensemble.GradientBoostingClassifier(),
    ensemble.RandomForestClassifier(),

    #Gaussian Processes
    gaussian_process.GaussianProcessClassifier(),
    
    #GLM
    linear_model.LogisticRegressionCV(),
    linear_model.PassiveAggressiveClassifier(),
    linear_model.RidgeClassifierCV(),
    linear_model.SGDClassifier(),
    linear_model.Perceptron(),
    
    #Navies Bayes
    naive_bayes.BernoulliNB(),
    naive_bayes.GaussianNB(),
    
    #Nearest Neighbor
    neighbors.KNeighborsClassifier(),
    
    #SVM
    svm.SVC(probability=True),
    svm.NuSVC(probability=True),
    svm.LinearSVC(),
    
    #Trees    
    tree.DecisionTreeClassifier(),
    tree.ExtraTreeClassifier(),
    
    #Discriminant Analysis
    discriminant_analysis.LinearDiscriminantAnalysis(),
    discriminant_analysis.QuadraticDiscriminantAnalysis(),

    #Neural Network
    neural_network.MLPClassifier(),
    
    #xgboost
    XGBClassifier(),
    XGBRFClassifier()
]
```

```python
# In[18]:


MLA_compare=["Algorithm_Name","Parameters", "Mean_Accuracy","Standard_Deviation"]
MLA_compare=pd.DataFrame(columns=MLA_compare)
```

```python
# In[19]:


from sklearn.model_selection import cross_val_score

index=0
for algo in MLA:
    clf=algo
    clf.fit(X_train,Y)
    cv=cross_val_score(estimator=algo,X=X_train,y=Y,cv=10,n_jobs=-1)
    cv_mean=cv.mean()
    cv_std=cv.std()
    MLA_compare.loc[index,"Algorithm_Name"]=algo.__class__.__name__
    MLA_compare.loc[index,"Parameters"]=str(algo.get_params())
    MLA_compare.loc[index,"Mean_Accuracy"]=cv_mean
    MLA_compare.loc[index,"Standard_Deviation"]=cv_std
    index+=1
```
